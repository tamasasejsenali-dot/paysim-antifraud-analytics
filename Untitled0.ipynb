```python
from google.colab import drive
import pandas as pd
import numpy as np
import duckdb

# Подключаем Google Диск
drive.mount('/content/drive')

# Настраиваем красивый вывод чисел в Pandas (без научной нотации вроде 1.2e+06)
pd.set_option('display.float_format', '{:,.2f}'.format)
pd.set_option('display.max_columns', None)

# Укажите ваш путь к распакованному CSV-файлу
file_path = '/content/drive/MyDrive/Portfolio_Fraud_Analysis/PS_20174392719_1491204439457_log.csv'

# Загружаем данные
df = pd.read_csv(file_path)

print(f"Размер датасета: {df.shape[0]:,} строк и {df.shape[1]} колонок\n")

# Проверяем пропуски и дубликаты
print("--- Количество пропусков по колонкам ---")
print(df.isnull().sum())
print(f"\nПолных дубликатов строк: {df.duplicated().sum()}")

# Смотрим первые 5 строк и типы данных
display(df.head())
df.info()

query_by_type = """
SELECT
    type AS transaction_type,
    COUNT(*) AS total_tx,
    SUM(isFraud) AS fraud_tx,
    ROUND(SUM(isFraud) * 100.0 / COUNT(*), 4) AS fraud_rate_pct,
    ROUND(SUM(amount), 2) AS total_volume,
    ROUND(SUM(CASE WHEN isFraud = 1 THEN amount ELSE 0 END), 2) AS fraud_volume,
    ROUND(AVG(CASE WHEN isFraud = 1 THEN amount END), 2) AS avg_fraud_amount,
    ROUND(AVG(CASE WHEN isFraud = 0 THEN amount END), 2) AS avg_legit_amount
FROM df
GROUP BY type
ORDER BY fraud_tx DESC, total_tx DESC;
"""

summary_by_type = duckdb.query(query_by_type).df()
display(summary_by_type)

query_flagged = """
SELECT
    isFraud,
    isFlaggedFraud,
    COUNT(*) AS tx_count,
    ROUND(SUM(amount), 2) AS total_amount
FROM df
WHERE type IN ('TRANSFER', 'CASH_OUT')
GROUP BY isFraud, isFlaggedFraud
ORDER BY isFraud DESC, isFlaggedFraud DESC;
"""

display(duckdb.query(query_flagged).df())

query_hourly = """
SELECT
    step % 24 AS hour_of_day,
    COUNT(*) AS total_tx,
    SUM(CASE WHEN isFraud = 0 THEN 1 ELSE 0 END) AS legit_tx,
    SUM(isFraud) AS fraud_tx,
    ROUND(SUM(isFraud) * 100.0 / COUNT(*), 2) AS fraud_rate_pct
FROM df
WHERE type IN ('TRANSFER', 'CASH_OUT')
GROUP BY hour_of_day
ORDER BY hour_of_day;
"""

hourly_df = duckdb.query(query_hourly).df()
display(hourly_df)

query_zero_balance = """
SELECT
    isFraud,
    COUNT(*) AS total_tx,
    SUM(CASE WHEN oldbalanceOrg > 0 AND newbalanceOrig = 0 THEN 1 ELSE 0 END) AS zeroed_out_tx,
    ROUND(
        SUM(CASE WHEN oldbalanceOrg > 0 AND newbalanceOrig = 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS zeroed_out_pct,
    SUM(CASE WHEN amount = oldbalanceOrg AND oldbalanceOrg > 0 THEN 1 ELSE 0 END) AS exact_full_amount_tx,
    ROUND(
        SUM(CASE WHEN amount = oldbalanceOrg AND oldbalanceOrg > 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*),
        2
    ) AS exact_full_amount_pct
FROM df
WHERE type IN ('TRANSFER', 'CASH_OUT')
GROUP BY isFraud;
"""

zero_balance_df = duckdb.query(query_zero_balance).df()
display(zero_balance_df)

query_balance_errors = """
WITH balance_checks AS (
    SELECT
        isFraud,
        amount,
        ROUND(oldbalanceOrg - amount - newbalanceOrig, 2) AS orig_error,
        ROUND(oldbalanceDest + amount - newbalanceDest, 2) AS dest_error
    FROM df
    WHERE type IN ('TRANSFER', 'CASH_OUT')
)
SELECT
    isFraud,
    COUNT(*) AS total_tx,
    -- Ошибка на стороне отправителя (деньги списались неверно или баланс остался 0)
    ROUND(SUM(CASE WHEN orig_error != 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS orig_error_pct,
    -- Ошибка на стороне получателя (деньги зачислились неверно или баланс не изменился)
    ROUND(SUM(CASE WHEN dest_error != 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS dest_error_pct,
    -- Ситуация, когда у получателя баланс был 0 до и остался 0 после крупного перевода
    ROUND(SUM(CASE WHEN dest_error = amount THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS dest_unchanged_pct
FROM balance_checks
GROUP BY isFraud;
"""

balance_errors_df = duckdb.query(query_balance_errors).df()
display(balance_errors_df)
```

```python
from google.colab import drive
import pandas as pd
import duckdb

# Подключаем Диск и заново загружаем таблицу в память
drive.mount('/content/drive')
file_path = '/content/drive/MyDrive/Portfolio_Fraud_Analysis/PS_20174392719_1491204439457_log.csv'
df = pd.read_csv(file_path)

# Выполняем расчет эффективности антифрод-правил
query_antifraud_rules = """
WITH features AS (
    SELECT
        isFraud,
        isFlaggedFraud,
        amount,
        CASE WHEN type IN ('TRANSFER', 'CASH_OUT') THEN 1 ELSE 0 END AS is_risky_type,
        CASE WHEN (step % 24) BETWEEN 0 AND 7 THEN 1 ELSE 0 END AS is_night,
        CASE WHEN oldbalanceOrg > 0 AND newbalanceOrig = 0 AND amount = oldbalanceOrg THEN 1 ELSE 0 END AS is_zeroed_out,
        CASE WHEN ROUND(oldbalanceOrg - amount - newbalanceOrig, 2) = 0 THEN 1 ELSE 0 END AS is_exact_debit
    FROM df
),
rules_applied AS (
    SELECT
        isFraud,
        amount,
        isFlaggedFraud AS rule_0_current_bank,
        CASE WHEN is_risky_type = 1 AND is_night = 1 AND is_zeroed_out = 1 THEN 1 ELSE 0 END AS rule_1_night_zero,
        CASE WHEN is_risky_type = 1 AND is_zeroed_out = 1 AND is_exact_debit = 1 THEN 1 ELSE 0 END AS rule_2_zero_exact_debit,
        CASE WHEN (is_risky_type = 1 AND is_zeroed_out = 1 AND is_exact_debit = 1)
               OR (is_risky_type = 1 AND is_night = 1 AND amount > 200000) THEN 1 ELSE 0 END AS rule_3_combined
    FROM features
),
totals AS (
    SELECT
        SUM(isFraud) AS total_fraud_tx,
        SUM(CASE WHEN isFraud = 1 THEN amount ELSE 0 END) AS total_fraud_sum,
        SUM(CASE WHEN isFraud = 0 THEN 1 ELSE 0 END) AS total_legit_tx
    FROM rules_applied
)
SELECT
    '0. Текущая система банка (isFlaggedFraud)' AS rule_name,
    SUM(CASE WHEN rule_0_current_bank = 1 AND isFraud = 1 THEN 1 ELSE 0 END) AS caught_fraud_tx,
    ROUND(SUM(CASE WHEN rule_0_current_bank = 1 AND isFraud = 1 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_fraud_tx), 2) AS recall_pct,
    ROUND(SUM(CASE WHEN rule_0_current_bank = 1 AND isFraud = 1 THEN amount ELSE 0 END) * 100.0 / MAX(t.total_fraud_sum), 2) AS saved_money_pct,
    SUM(CASE WHEN rule_0_current_bank = 1 AND isFraud = 0 THEN 1 ELSE 0 END) AS false_positives_tx,
    ROUND(SUM(CASE WHEN rule_0_current_bank = 1 AND isFraud = 0 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_legit_tx), 4) AS false_positive_rate_pct
FROM rules_applied CROSS JOIN totals t

UNION ALL

SELECT
    '1. Правило: Ночь + Обнуление баланса',
    SUM(CASE WHEN rule_1_night_zero = 1 AND isFraud = 1 THEN 1 ELSE 0 END),
    ROUND(SUM(CASE WHEN rule_1_night_zero = 1 AND isFraud = 1 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_fraud_tx), 2),
    ROUND(SUM(CASE WHEN rule_1_night_zero = 1 AND isFraud = 1 THEN amount ELSE 0 END) * 100.0 / MAX(t.total_fraud_sum), 2),
    SUM(CASE WHEN rule_1_night_zero = 1 AND isFraud = 0 THEN 1 ELSE 0 END),
    ROUND(SUM(CASE WHEN rule_1_night_zero = 1 AND isFraud = 0 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_legit_tx), 4)
FROM rules_applied CROSS JOIN totals t

UNION ALL

SELECT
    '2. Правило: Обнуление + Точное списание (24/7)',
    SUM(CASE WHEN rule_2_zero_exact_debit = 1 AND isFraud = 1 THEN 1 ELSE 0 END),
    ROUND(SUM(CASE WHEN rule_2_zero_exact_debit = 1 AND isFraud = 1 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_fraud_tx), 2),
    ROUND(SUM(CASE WHEN rule_2_zero_exact_debit = 1 AND isFraud = 1 THEN amount ELSE 0 END) * 100.0 / MAX(t.total_fraud_sum), 2),
    SUM(CASE WHEN rule_2_zero_exact_debit = 1 AND isFraud = 0 THEN 1 ELSE 0 END),
    ROUND(SUM(CASE WHEN rule_2_zero_exact_debit = 1 AND isFraud = 0 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_legit_tx), 4)
FROM rules_applied CROSS JOIN totals t

UNION ALL

SELECT
    '3. Комбинированное правило (Обнуление + Ночной фильтр)',
    SUM(CASE WHEN rule_3_combined = 1 AND isFraud = 1 THEN 1 ELSE 0 END),
    ROUND(SUM(CASE WHEN rule_3_combined = 1 AND isFraud = 1 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_fraud_tx), 2),
    ROUND(SUM(CASE WHEN rule_3_combined = 1 AND isFraud = 1 THEN amount ELSE 0 END) * 100.0 / MAX(t.total_fraud_sum), 2),
    SUM(CASE WHEN rule_3_combined = 1 AND isFraud = 0 THEN 1 ELSE 0 END),
    ROUND(SUM(CASE WHEN rule_3_combined = 1 AND isFraud = 0 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_legit_tx), 4)
FROM rules_applied CROSS JOIN totals t;
"""

rules_comparison_df = duckdb.query(query_antifraud_rules).df()
display(rules_comparison_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import duckdb
import os

# Создаем папку для сохранения графиков на Google Диске
output_dir = '/content/drive/MyDrive/Portfolio_Fraud_Analysis/charts'
os.makedirs(output_dir, exist_ok=True)

# Настраиваем общий минималистичный стиль графиков
sns.set_theme(style="whitegrid")
plt.rcParams['font.sans-serif'] = 'DejaVu Sans'
plt.rcParams['axes.edgecolor'] = '#CCCCCC'
plt.rcParams['axes.linewidth'] = 0.8

# ==============================================================================
# ГРАФИК 1: Суточная динамика доли мошенничества (Fraud Rate по часам суток)
# ==============================================================================
hourly_df = duckdb.query("""
    SELECT
        step % 24 AS hour_of_day,
        ROUND(SUM(isFraud) * 100.0 / COUNT(*), 2) AS fraud_rate_pct
    FROM df
    WHERE type IN ('TRANSFER', 'CASH_OUT')
    GROUP BY hour_of_day
    ORDER BY hour_of_day
""").df()

fig, ax = plt.subplots(figsize=(12, 5))

# Подсвечиваем ночные часы (0-7) красным цветом, дневные — нейтральным серо-синим
colors_hourly = ['#D9534F' if h <= 7 else '#5B84B1' for h in hourly_df['hour_of_day']]
bars = sns.barplot(data=hourly_df, x='hour_of_day', y='fraud_rate_pct', palette=colors_hourly, ax=ax)

# Выделяем ночную зону фоном и подписью
ax.axvspan(-0.5, 7.5, color='#D9534F', alpha=0.08, label='Ночная зона повышенного риска (00:00 – 07:00)')

for bar in bars.patches:
    height = bar.get_height()
    if height > 2:  # Подписываем только заметные столбцы, чтобы не засорять график
        ax.annotate(f'{height:.1f}%',
                    xy=(bar.get_x() + bar.get_width() / 2, height),
                    xytext=(0, 4), textcoords="offset points",
                    ha='center', va='bottom', fontsize=9, fontweight='bold', color='#333333')

ax.set_title('Доля мошеннических операций по часам суток (TRANSFER и CASH_OUT)', fontsize=14, fontweight='bold', pad=15)
ax.set_xlabel('Час суток (0 = полночь)', fontsize=11)
ax.set_ylabel('Доля фрода (Fraud Rate, %)', fontsize=11)
ax.set_ylim(0, hourly_df['fraud_rate_pct'].max() * 1.15)
ax.legend(frameon=True, facecolor='white', fontsize=10)
sns.despine()

plt.tight_layout()
plt.savefig(f'{output_dir}/1_hourly_fraud_rate.png', dpi=300)
plt.show()

# ==============================================================================
# ГРАФИК 2: Поведенческий паттерн — полное обнуление баланса отправителя
# ==============================================================================
zero_df = duckdb.query("""
    SELECT
        CASE WHEN isFraud = 1 THEN 'Мошенники (isFraud = 1)' ELSE 'Обычные клиенты (isFraud = 0)' END AS client_type,
        ROUND(SUM(CASE WHEN oldbalanceOrg > 0 AND newbalanceOrig = 0 THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 1) AS zeroed_pct
    FROM df
    WHERE type IN ('TRANSFER', 'CASH_OUT')
    GROUP BY isFraud
""").df()

fig, ax = plt.subplots(figsize=(8, 5))
bars2 = sns.barplot(data=zero_df, x='client_type', y='zeroed_pct', palette=['#5B84B1', '#D9534F'], width=0.5, ax=ax)

for bar in bars2.patches:
    height = bar.get_height()
    ax.annotate(f'{height:.1f}%',
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 5), textcoords="offset points",
                ha='center', va='bottom', fontsize=12, fontweight='bold', color='#222222')

ax.set_title('Доля транзакций с полным обнулением счета отправителя', fontsize=13, fontweight='bold', pad=15)
ax.set_xlabel('')
ax.set_ylabel('Доля операций с обнулением баланса (%)', fontsize=11)
ax.set_ylim(0, 112)
sns.despine()

plt.tight_layout()
plt.savefig(f'{output_dir}/2_zero_balance_comparison.png', dpi=300)
plt.show()

# ==============================================================================
# ГРАФИК 3: Сравнение эффективности антифрод-правил (Recall vs False Positives)
# ==============================================================================
rules_df = duckdb.query("""
    WITH features AS (
        SELECT
            isFraud,
            isFlaggedFraud,
            amount,
            CASE WHEN type IN ('TRANSFER', 'CASH_OUT') THEN 1 ELSE 0 END AS is_risky_type,
            CASE WHEN (step % 24) BETWEEN 0 AND 7 THEN 1 ELSE 0 END AS is_night,
            CASE WHEN oldbalanceOrg > 0 AND newbalanceOrig = 0 AND amount = oldbalanceOrg THEN 1 ELSE 0 END AS is_zeroed_out,
            CASE WHEN ROUND(oldbalanceOrg - amount - newbalanceOrig, 2) = 0 THEN 1 ELSE 0 END AS is_exact_debit
        FROM df
    ),
    totals AS (
        SELECT SUM(isFraud) AS total_fraud, SUM(CASE WHEN isFraud = 0 THEN 1 ELSE 0 END) AS total_legit FROM features
    )
    SELECT 'Текущий фильтр банка' AS rule,
           ROUND(SUM(CASE WHEN isFlaggedFraud = 1 AND isFraud = 1 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_fraud), 2) AS recall,
           ROUND(SUM(CASE WHEN isFlaggedFraud = 1 AND isFraud = 0 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_legit), 3) AS fpr
    FROM features CROSS JOIN totals t
    UNION ALL
    SELECT 'Правило 1: Ночь + Обнуление',
           ROUND(SUM(CASE WHEN is_risky_type=1 AND is_night=1 AND is_zeroed_out=1 AND isFraud=1 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_fraud), 2),
           ROUND(SUM(CASE WHEN is_risky_type=1 AND is_night=1 AND is_zeroed_out=1 AND isFraud=0 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_legit), 3)
    FROM features CROSS JOIN totals t
    UNION ALL
    SELECT 'Правило 2: Обнуление + Точное списание (24/7)',
           ROUND(SUM(CASE WHEN is_risky_type=1 AND is_zeroed_out=1 AND is_exact_debit=1 AND isFraud=1 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_fraud), 2),
           ROUND(SUM(CASE WHEN is_risky_type=1 AND is_zeroed_out=1 AND is_exact_debit=1 AND isFraud=0 THEN 1 ELSE 0 END) * 100.0 / MAX(t.total_legit), 3)
    FROM features CROSS JOIN totals t
""").df()

fig, ax = plt.subplots(figsize=(10, 5))
palette_rules = ['#95A5A6', '#F0AD4E', '#2E8B57']
bars3 = sns.barplot(data=rules_df, y='rule', x='recall', palette=palette_rules, width=0.55, ax=ax)

for idx, bar in enumerate(bars3.patches):
    width = bar.get_width()
    fpr_val = rules_df.loc[idx, 'fpr']
    ax.annotate(f'Поймано фрода: {width:.1f}%  (ложных блокировок: {fpr_val:.3f}%)',
                xy=(width, bar.get_y() + bar.get_height() / 2),
                xytext=(6, 0), textcoords="offset points",
                ha='left', va='center', fontsize=10, fontweight='bold', color='#222222')

ax.set_title('Сравнение эффективности антифрод-правил (Доля пойманного фрода, Recall)', fontsize=13, fontweight='bold', pad=15)
ax.set_xlabel('Пойманные мошеннические транзакции (Recall, %)', fontsize=11)
ax.set_ylabel('')
ax.set_xlim(0, 145)  # Запас справа для текстовых подписей
sns.despine()

plt.tight_layout()
plt.savefig(f'{output_dir}/3_antifraud_rules_comparison.png', dpi=300)
plt.show()

print(f"\nВсе 3 графика успешно сохранены в папку: {output_dir}")
```

```python
import duckdb

# Путь для сохранения итоговой витрины на Google Диск
mart_path = '/content/drive/MyDrive/Portfolio_Fraud_Analysis/paysim_bi_datamart.csv'

query_datamart = f"""
COPY (
    WITH enriched_tx AS (
        SELECT
            -- Временные измерения
            CAST((step - 1) / 24 AS INTEGER) + 1 AS sim_day,
            step % 24 AS hour_of_day,
            CASE
                WHEN (step % 24) BETWEEN 0 AND 7 THEN 'Ночь (00:00 - 07:00)'
                ELSE 'День (08:00 - 23:00)'
            END AS time_period,

            -- Бизнес-измерения
            type AS transaction_type,
            CASE
                WHEN amount < 10000 THEN '1. До 10 тыс.'
                WHEN amount < 100000 THEN '2. 10 тыс. – 100 тыс.'
                WHEN amount < 500000 THEN '3. 100 тыс. – 500 тыс.'
                ELSE '4. Более 500 тыс.'
            END AS amount_bucket,

            -- Поведенческие флаги
            CASE
                WHEN oldbalanceOrg > 0 AND newbalanceOrig = 0 THEN 'Обнуление счета'
                ELSE 'Обычный остаток'
            END AS balance_behavior,

            -- Срабатывание нашего лучшего антифрод-правила (Правило 2: Рисковый тип + Обнуление + Точное списание)
            CASE
                WHEN type IN ('TRANSFER', 'CASH_OUT')
                     AND oldbalanceOrg > 0
                     AND newbalanceOrig = 0
                     AND amount = oldbalanceOrg
                     AND ROUND(oldbalanceOrg - amount - newbalanceOrig, 2) = 0
                THEN 1 ELSE 0
            END AS new_rule_triggered,

            isFraud,
            isFlaggedFraud,
            amount
        FROM df
    )
    SELECT
        sim_day,
        hour_of_day,
        time_period,
        transaction_type,
        amount_bucket,
        balance_behavior,

        -- Агрегированные метрики (факты)
        COUNT(*) AS total_tx_count,
        ROUND(SUM(amount), 2) AS total_tx_volume,

        SUM(isFraud) AS fraud_tx_count,
        ROUND(SUM(CASE WHEN isFraud = 1 THEN amount ELSE 0 END), 2) AS fraud_tx_volume,

        SUM(CASE WHEN isFraud = 0 THEN 1 ELSE 0 END) AS legit_tx_count,

        -- Метрики старой системы банка
        SUM(CASE WHEN isFlaggedFraud = 1 AND isFraud = 1 THEN 1 ELSE 0 END) AS old_rule_caught_fraud,

        -- Метрики нашего нового правила
        SUM(CASE WHEN new_rule_triggered = 1 AND isFraud = 1 THEN 1 ELSE 0 END) AS new_rule_caught_fraud,
        ROUND(SUM(CASE WHEN new_rule_triggered = 1 AND isFraud = 1 THEN amount ELSE 0 END), 2) AS new_rule_saved_volume,
        SUM(CASE WHEN new_rule_triggered = 1 AND isFraud = 0 THEN 1 ELSE 0 END) AS new_rule_false_positives

    FROM enriched_tx
    GROUP BY
        sim_day,
        hour_of_day,
        time_period,
        transaction_type,
        amount_bucket,
        balance_behavior
    ORDER BY sim_day, hour_of_day, transaction_type
) TO '{mart_path}' (HEADER, DELIMITER ',');
"""

duckdb.query(query_datamart)

# Проверяем, что файл создался и смотрим его размер
mart_preview = duckdb.query(f"SELECT * FROM '{mart_path}' LIMIT 5").df()
row_count = duckdb.query(f"SELECT COUNT(*) AS cnt FROM '{mart_path}'").df()['cnt'][0]

print(f"Витрина успешно сохранена: {mart_path}")
print(f"Итоговое количество строк для BI: {row_count:,}")
display(mart_preview)
```
